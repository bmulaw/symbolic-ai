## Q4 [Total: 9 points] 

### Q4a. [3 points] 

List the main benefits and limitations of Symbolic AI.

Symbolic AI works without oceans of data so it is simplier than data-driven AI. Symbolic AI are also easier to decipher and understand how a conclusion has been reached (by looking at it's propositional logic and decisions at each step). However, Symbolic AI can be difficult to build because there can be a ton of symbols and rules that needs to be set up for it to work in some scenarios (i.e. hard to scale). This also means that one Symbolic AI cannot be generalized to make good decisions under different environments.

### Q4b. [3 points] 

List the main benefits and limitation(s) of Neural Networks.

Neural Networks (NN) works well with high volume of data, even datasets containing some faulty data can train NN without distributing it. NN can also learn from it's previous mistakes and can perform better by seeing more examples of past events. However, NN can be very complex and difficult to decipher (hidden layers make it difficult to understand why some decision were made). Also, NN can be time consuming to run especially if you rely on slower hardware - thus being less practical solution for some.
